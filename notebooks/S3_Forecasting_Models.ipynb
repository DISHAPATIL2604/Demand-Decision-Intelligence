# Stage S3: Forecasting Models (Weeks 5–6)
**Demand-Decision-Intelligence System**

This notebook covers the complete workflow for **Stage S3: Forecasting Models**:
1. **Time-Series Train/Validation Split**: Chronological split (`train < 2022-07-01`, `val >= 2022-07-01`)
2. **Baseline Models**: Naive (Lag-1), Seasonal Naive (Lag-7)
3. **Moving Averages**: 7-day, 14-day, and 28-day rolling averages
4. **Prophet Model**: Facebook Prophet with weekly seasonality
5. **Machine Learning Models**: Ridge Regression & HistGradientBoosting (GBT)
6. **Model Evaluation & Comparison**: MAE, RMSE, WAPE Metrics and CSV Deliverables Export

```python
import os
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error
from prophet import Prophet

# Load Model Comparison Deliverable
reports_dir = Path("../reports")
df_comp = pd.read_csv(reports_dir / "model_comparison.csv")
df_comp
```

## 1. Model Comparison Summary
Below is the benchmark table comparing all 8 evaluated models across MAE, RMSE, and WAPE (%).

```python
fig, ax = plt.subplots(figsize=(10, 5))
df_comp_sorted = df_comp.sort_values(by="WAPE_pct", ascending=True)
ax.barh(df_comp_sorted["model"], df_comp_sorted["WAPE_pct"], color="#2b5c8f")
ax.set_xlabel("Weighted Absolute Percentage Error (WAPE %)")
ax.set_title("Stage S3: Forecasting Models WAPE (%) Comparison")
ax.grid(axis="x", linestyle="--", alpha=0.7)
plt.tight_layout()
plt.show()
```
